```python
# !pip install transforms3d
import numpy as np
from transforms3d.euler import quat2euler, euler2quat

# Quaternion format: [w, x, y, z]
quat = [0.7071068, 0.0, -0.7071068, 0.0]

# Convert quaternion to Euler angles (in radians), XYZ axes order
euler_angles = quat2euler(quat, axes='sxyz')

print("Euler angles (radians):", euler_angles)
print("Euler angles (degrees):", np.degrees(euler_angles))

euler_angles = [0, np.pi/2, 0]

# Convert Euler angles to quaternion (w, x, y, z), XYZ axes order
quat = euler2quat(*euler_angles, axes='sxyz')

print("Quaternion [w,x,y,z]:", quat)
```
